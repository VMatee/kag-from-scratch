# Knowledge graph evidence paths

This repository models manually curated subject-relation-object statements with source provenance. A directed graph search retrieves paths between two entities. This is **not** a graph-building LLM.

```python
from pathlib import Path
from kag_lab import KnowledgeGraph, load_triples, render_path

path = Path("../data/triples.jsonl")
if not path.exists():
    path = Path("data/triples.jsonl")
triples = load_triples(path)
graph = KnowledgeGraph(triples)
print("Triples:", len(triples))
```

## Trace a three-hop explanation

A path is accepted only when every edge exists in the source graph.

```python
paths = graph.find_paths("GitHub", "code review", max_hops=3)
for i, path in enumerate(paths, 1):
    print("Path", i)
    print(render_path(path))
```

## Change the hop budget

```python
assert graph.find_paths("GitHub", "code review", max_hops=2) == []
print("2 hops: none; 3 hops:", len(paths))
```
